# Pipeline API example

Prepare, run and analyze a G-quadruplex folding simulation with
`FoldingPipeline`. Edit the inputs and working directory, then run the
stages in order. **Choose one config loader: dictionary or JSON.**

Use the kernel with `g4_builder` installed; see the
[installation instructions](../README.md#installation-and-environment).
Activate GROMACS before starting the kernel. Minimization and TPR
preparation require GROMACS; running the simulation additionally needs
PLUMED support. Choose a new working directory for a new simulation.

````python
import g4_builder as g4
````

## 1. Define inputs and the working directory

Set the simulation inputs and choose a new working directory on your machine. The dictionary is used by configuration option A; option B reads the settings from JSON instead. Supply the working directory separately in either case.

````python
# 1. Define simulation inputs and choose a new working directory.
# Option A uses this dictionary; option B reads the settings from JSON.
# working_dir is supplied separately for both options.
input_dict = {
    # Mandatory 
    "sequence": "GGGTAGGGCGGCGGGGCAGGGT",
    "GT_resids": {
        "GT_1": [11, 1, 2],
        "GT_2": [8, 7, 6],
        "GT_3": [19, 20, 21],
        "GT_4": [15, 14, 13]
    },
    "polarity_pattern": "ssa",
    "helicity": "RH",
    "force_field": 5,
    "gromacs_executable": "gmx",

    # Optional (use defaults if not provided)
    "after_folding_steps": 2500000,
    "schedule": {
        "gt_initial_steps": 400000,
        "core_start": 1000000,
        "core_steps": 3000000,
        "tetrads_steps": 3000000
    }
}

working_dir = "example_results/pipeline_api"
````

## 2. Construct file paths

Use the package filename defaults to construct input and output paths.

````python
# 2. Construct the input/output paths using the package filename defaults.
paths = g4.RunPaths(working_dir=working_dir)
````

## 3. Configure logging

Write detailed GROMACS output to the log and show brief console messages.

````python
# 3. Configure detailed file logging and brief console messages.
g4.configure_logging(paths.log)
````

## 4. Select a JSON input file (option B only)

For the JSON option, place an existing input.json in the working directory or change input_file to its location. It contains simulation settings, without working_dir. It is separate from the generated run.json metadata.

````python
# 4. Option B only: select an existing simulation-settings JSON file.
# Change this path if input.json is stored elsewhere.
input_file = paths.working_dir / "input.json"
````

## 5A. Create the config from a dictionary

**Choose this cell or option B below.** Missing optional settings use their defaults.

````python
# 5. Option A: load settings from the dictionary.
# Run this cell OR the JSON-loading cell below.
config = g4.SimulationConfig.from_dict(
    values=input_dict,
    working_dir=paths.working_dir
)
````

## 5B. Create the config from JSON

**Alternative to option A.** The JSON loader uses the same settings names and also receives the working directory separately.

````python
# 5. Option B: load settings from the JSON input file.
# Run this cell INSTEAD OF the dictionary-loading cell above.
config = g4.SimulationConfig.from_json(
    path=input_file,
    working_dir=paths.working_dir
)
````

## 6. Create the pipeline

Construct the pipeline from the chosen config. This does not start GROMACS.

````python
# 6. Create the pipeline using the config selected above.
fold_pipe = g4.FoldingPipeline(config)
````

## 7. Prepare strand, references and protocol

Build and save the inputs that do not require GROMACS. The pipeline writes run.json metadata after each successful stage.

````python
# 7. Build the strand, references and PLUMED protocol without GROMACS.
# Successful pipeline stages save their state in run.json.
fold_pipe.prepare()
````

## 8. Generate topology and minimize

Run topology preparation, check reference atom numbering, add the box and minimize the structure. GROMACS is required.

````python
# 8. Generate topology, remap changed reference serials, add a box and minimize.
# This stage requires GROMACS.
fold_pipe.minimize()
````

## 9. Prepare the simulation TPR

Render the simulation MDP and prepare the TPR using the minimized structure. The total steps include the folding protocol and configured extra steps.

````python
# 9. Prepare the simulation MDP and TPR from the minimized structure.
fold_pipe.prepare_simulation()
````

## 10. Run the simulation locally (optional)

Run this cell only with a **PLUMED-enabled GROMACS** installation. Otherwise skip it and run the prepared simulation on another machine as described below.

````python
# 10. Optional: run locally with a PLUMED-enabled GROMACS installation.
# Otherwise skip this cell, run remotely and copy results back before extraction.
fold_pipe.run()
````

### Run elsewhere if local GROMACS lacks PLUMED

After `prepare_simulation()`, copy the entire working directory to
the other machine, keeping the TPR, PLUMED protocol, references and
run.json together. From inside that directory, use a PLUMED-enabled
GROMACS installation. With the default filenames:

```bash
gmx mdrun -deffnm smd -cpi -v -plumed plumed_protocol.dat -c smd.pdb
```

Use your site's job-submission system if required. Copy COLVAR and
smd.xtc back into the local working directory and
retain the TPR used for the simulation. Skip `fold_pipe.run()` and
continue with extraction. Local GROMACS is needed for extraction.

After a kernel restart, rerun the import and first input cell to set
`working_dir`. Restore the prepared pipeline instead of repeating
preparation:

```python
fold_pipe = g4.FoldingPipeline.from_directory(working_dir)
paths = fold_pipe.paths
```

## 11. Extract the lowest-RMSD frames

Select post-protocol frames using the full-core RMSD label and extract them to PDB files. An explicit trajectory path allows this step after a remote run as well.

````python
# 11. Select and extract the lowest-RMSD frames after the folding protocol.
# Explicit paths also support remote runs after copying results back.
fold_pipe.extract(trajectories=paths.trajectory)
````
